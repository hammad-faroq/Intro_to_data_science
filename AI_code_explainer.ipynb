and responds with an explanation. This is a tool that you will be able to use yourself during the course!

````python
# imports
from openai import OpenAI
from IPython.display import display, Markdown, update_display
````

````python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2:1b'
````

````python
# set up environment
openai = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
````

````python
# /*//////# here is the question; type over this to ask something new
code = input("Enter the code which you want to analyse using my AI: ")

system_prompt = """
You are a code tutor for me who will summarize and explain what the code is doing.
The code will mostly be in Python. Please keep your explanation short and to the point.
"""

user_prompt = f"""
Please explain what this code does and why:
\"\"\"{code}\"\"\"
"""
````

````python
# Get Llama 3.2 to answer
response1=openai.chat.completions.create(
    model=MODEL_LLAMA,
    messages=[{"role":"system","content":system_prompt},
            {"role":"user","content":user_prompt}],
    stream=True
)
response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in response1:
    response += chunk.choices[0].delta.content or ''
    response = response.replace("```","").replace("markdown", "")
    update_display(Markdown(response), display_id=display_handle.display_id)
# print(response.choices[0].message.content)
````
