```python
import gradio as gr
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

# Load an open-source model (change this to quantized if needed)
MODEL_NAME = "Salesforce/codegen-350M-mono"  # small, just for demo
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(MODEL_NAME)

# Pipeline
generator = pipeline("text-generation", model=model, tokenizer=tokenizer)

def add_comments_and_docstrings(code):
    prompt = f"""
            You are a Python expert. Add inline comments to each line of the following Python code to explain what it does.

            ⚠️ IMPORTANT:
            - Do NOT change or reformat any line of the code.
            - Do NOT add new lines of code or extra print statements.
            - Only add comments using the '#' symbol at the end or above the lines.
            - Keep the code logic exactly as it is.

            Here is the code:
            {code}
            """

    output = generator(prompt, max_new_tokens=512, do_sample=True, temperature=0.7)[0]["generated_text"]
    return output.replace(prompt, "").strip()

# Gradio App
iface = gr.Interface(
    fn=add_comments_and_docstrings,
    inputs=gr.Textbox(label="Paste Python Code", lines=15, placeholder="def add(a, b):\n    return a + b"),
    outputs=gr.Textbox(label="Code with Comments + Docstrings", lines=15),
    title="🧠 Auto Commenter & Docstring Generator",
    description="Uses an open-source LLM to enhance your Python code by adding inline comments and docstrings."
)

iface.launch(share=True)
```
