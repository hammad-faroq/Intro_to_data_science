**IMPORT THE RELAVENT LIBRARIES**

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('df_preprocessed.csv')
df
```

```python
targets=np.where(df['Absenteeism Time in Hours']> df['Absenteeism Time in Hours'].median(),1,0)
targets
```

```python
df['Excessive Absenteesiam']=targets
```

```python
df.head()
```

```python

df=df.drop(['Daily Work Load Average'],axis=1)
df=df.drop(['Day of the Week'],axis=1)
df=df.drop(['Distance to Work'],axis=1)
df=df.drop(['Pets'],axis=1)
df=df.drop(['Absenteeism Time in Hours'],axis=1)
```

```python
df.head()
```

```python
unscaled_inputs=df.iloc[:,:-1]
unscaled_inputs
```

**STANDARIZE THE DATA**

```python
from sklearn.preprocessing import StandardScaler
```

```python
Absenteeism_scaler=StandardScaler()
Absenteeism_scaler.fit(unscaled_inputs)
```

```python
scaled_inputs=Absenteeism_scaler.transform(unscaled_inputs)
scaled_inputs
```

```python
scaled_inputs.shape
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(scaled_inputs, targets, test_size=0.2, random_state=20)
```

```python
x_train.shape, x_test.shape, y_train.shape
```

**logistic REGREsSION MODEL**

```python
from sklearn.linear_model import LogisticRegression
```

```python
reg=LogisticRegression()
reg.fit(x_train, y_train)
print("Train score:", reg.score(x_train, y_train))
print("Test score :", reg.score(x_test, y_test))
```

```python
reg.score(x_train, y_train)
```

```python
reg.score(x_test, y_test)
```

**SAVE HE MODEL**

```python
import pickle
with open('model','wb') as file:
    pickle.dump(reg,file)
with open('scaler', 'wb') as file:
    pickle.dump(Absenteeism_scaler, file)
```
